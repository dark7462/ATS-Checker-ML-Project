# 2. Match required technical skills and score
Run notebook 1 first. Full embedding vectors are compared with cosine similarity. Pairs above the 0.60 similarity floor earn calibrated credit, and each resume skill can be paired only once. Restart the kernel before running if `matcher.py` changed.


```python
from pathlib import Path
import json
from matcher import load_extractions, score

DATA_PATH = Path("output/extracted.json")
RESULT_PATH = Path("output/score.json")
data = load_extractions(DATA_PATH)
result = score(data["jd"], data["resume"])
RESULT_PATH.parent.mkdir(parents=True, exist_ok=True)
RESULT_PATH.write_text(json.dumps(result, indent=2, ensure_ascii=False), encoding="utf-8")
```

```python
print(f"Score: {result['score']:.2f}/100 ({result['points']}/{result['total']} required JD skills)")
print("\nVector matches:")
for row in result["matched"]: print(f"  {row['jd']} ↔ {row['resume']} (cosine {row['similarity']}, +{row['credit']} points)")
print("\nUnmatched required JD skills:")
for label in result["unmatched"]: print(f"  {label}")
print(f"\nSaved detailed result to {RESULT_PATH}")
```
