# 1. Extract required technical skills and store vectors
Put `JD.txt` and `resume.pdf` in `input/`. Run all cells. Only technical skills in the JD’s **Required Qualifications** section are scored; preferred qualifications and full sentences are excluded.

If this notebook was already open when `matcher.py` changed, use **Restart Kernel and Run All** so Python loads the current module.


```python
from pathlib import Path
from matcher import read_jd, focus_jd_text, read_resume_pdf, extract_items, embed_items, load_embedding_model, save_extractions

JD_PATH = Path("input/JD.txt")
RESUME_PATH = Path("input/resume.pdf")
OUTPUT_PATH = Path("output/extracted.json")
```

```python
embedding_model = load_embedding_model()
```

```python
jd_text = focus_jd_text(read_jd(JD_PATH))
resume_text = read_resume_pdf(RESUME_PATH)
jd_items = embed_items(extract_items(jd_text), embedding_model)
resume_items = embed_items(extract_items(resume_text, source="resume"), embedding_model)
if not jd_items:
    raise ValueError("Cannot extract any required JD technical skills. Check the input text and extraction rules.")
if not resume_items:
    raise ValueError("Cannot extract any resume items. Check the PDF text and extraction rules.")
save_extractions(jd_items, resume_items, OUTPUT_PATH)
print(f"Saved {len(jd_items)} JD items and {len(resume_items)} resume items to {OUTPUT_PATH}")
```

```python
print("Required JD technical skills:")
for item in jd_items:
    print(f"- {item['label']} → {item['canonical']}")
print("\nResume technical skills:")
for item in resume_items:
    print(f"- {item['label']} → {item['canonical']}")
```
